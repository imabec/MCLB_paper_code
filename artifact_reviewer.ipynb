```python
import pandas as pd
import numpy as np

df = pd.read_csv("reviewer_artifact_obshet_FIXED1.csv")

print(df.shape)
print(df["success"].value_counts(dropna=False))

clean = df[df["success"] == True].copy()
```

```python
stage_counts = (
    clean.groupby("period")
    .size()
    .reset_index(name="n_blocks")
)

stage_counts
```

```python
stage_params = (
    clean.groupby("period")
    .agg(
        n=("gamma_est", "size"),

        gamma_mean=("gamma_est", "mean"),
        gamma_median=("gamma_est", "median"),
        gamma_sd=("gamma_est", "std"),

        alpha_mean=("alpha_est", "mean"),
        alpha_median=("alpha_est", "median"),

        phi_mean=("phi_est", "mean"),
        phi_median=("phi_est", "median"),

        r_mean=("r_est", "mean"),
        r_median=("r_est", "median"),
    )
    .reset_index()
)

stage_params.round(4)
```

```python
obshet_summary = (
    clean.groupby("period")
    .agg(
        n=("delta_nll_obshet", "size"),

        mean_delta_nll_obshet=("delta_nll_obshet", "mean"),
        median_delta_nll_obshet=("delta_nll_obshet", "median"),

        mean_delta_bic_obshet=("delta_bic_obshet", "mean"),
        median_delta_bic_obshet=("delta_bic_obshet", "median"),

        joint_nll_win_rate=("joint_wins_obshet_nll", "mean"),
        joint_bic_win_rate=("joint_wins_obshet_bic", "mean"),
    )
    .reset_index()
)

obshet_summary[
    ["joint_nll_win_rate", "joint_bic_win_rate"]
] *= 100

obshet_summary.round(3)
```

```python
gam0_summary = (
    clean.groupby("period")
    .agg(
        n=("delta_nll_gamma0", "size"),

        mean_delta_nll_gamma0=("delta_nll_gamma0", "mean"),
        median_delta_nll_gamma0=("delta_nll_gamma0", "median"),

        mean_delta_bic_gamma0=("delta_bic_gamma0", "mean"),
        median_delta_bic_gamma0=("delta_bic_gamma0", "median")

    )
    .reset_index()
)

gam0_summary .round(3)
```

```python
gamma0_summary = (
    clean.groupby("period")
    .agg(
        n=("delta_nll_gamma0", "size"),

        mean_delta_nll_gamma0=("delta_nll_gamma0", "mean"),
        median_delta_nll_gamma0=("delta_nll_gamma0", "median"),

        mean_delta_bic_gamma0=("delta_bic_gamma0", "mean"),
        median_delta_bic_gamma0=("delta_bic_gamma0", "median"),

        joint_nll_win_rate=(
            "delta_nll_gamma0",
            lambda x: (x > 0).mean()
        ),

        joint_bic_win_rate=(
            "delta_bic_gamma0",
            lambda x: (x > 0).mean()
        ),
    )
    .reset_index()
)

# Convert win rates to percentages
gamma0_summary[
    ["joint_nll_win_rate", "joint_bic_win_rate"]
] *= 100

gamma0_summary.round(3)
```
