```python
import pandas as pd
import numpy as np

df = pd.read_csv("reviewer_model_selection_summary.csv")

print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
p500 = df[df["particles"] == 500].copy()

p500[
    [
        "mu_true",
        "alpha_true",
        "gamma_true",
        "phi_true",
        "r_true",
        "n",
        "mean_delta_nll",
        "mean_delta_nll_mcse",
        "mean_delta_bic",
        "mean_delta_bic_mcse",
        "joint_nll_win_rate",
        "joint_bic_win_rate",
    ]
].round(4)
```

```python
mcse_500_summary = pd.DataFrame({
    "quantity": ["Delta NLL", "Delta BIC"],
    "min_MCSE": [
        p500["mean_delta_nll_mcse"].min(),
        p500["mean_delta_bic_mcse"].min(),
    ],
    "median_MCSE": [
        p500["mean_delta_nll_mcse"].median(),
        p500["mean_delta_bic_mcse"].median(),
    ],
    "mean_MCSE": [
        p500["mean_delta_nll_mcse"].mean(),
        p500["mean_delta_bic_mcse"].mean(),
    ],
    "max_MCSE": [
        p500["mean_delta_nll_mcse"].max(),
        p500["mean_delta_bic_mcse"].max(),
    ]
})

mcse_500_summary.round(4)
```

```python
p500["delta_nll_signal_to_mcse"] = (
    np.abs(p500["mean_delta_nll"]) /
    p500["mean_delta_nll_mcse"]
)

p500["delta_bic_signal_to_mcse"] = (
    np.abs(p500["mean_delta_bic"]) /
    p500["mean_delta_bic_mcse"]
)

p500[
    [
        "alpha_true",
        "gamma_true",
        "mean_delta_nll",
        "mean_delta_nll_mcse",
        "delta_nll_signal_to_mcse",
        "mean_delta_bic",
        "mean_delta_bic_mcse",
        "delta_bic_signal_to_mcse",
    ]
].round(3)
```

```python
print(
    "Median |Delta NLL| / MCSE:",
    p500["delta_nll_signal_to_mcse"].median()
)

print(
    "Median |Delta BIC| / MCSE:",
    p500["delta_bic_signal_to_mcse"].median()
)
```

```python
particle_summary = (
    df.groupby("particles")
    .agg(
        n_settings=("mean_delta_nll", "size"),

        mean_delta_nll=("mean_delta_nll", "mean"),
        median_delta_nll=("mean_delta_nll", "median"),
        median_delta_nll_mcse=("mean_delta_nll_mcse", "median"),

        mean_delta_bic=("mean_delta_bic", "mean"),
        median_delta_bic=("mean_delta_bic", "median"),
        median_delta_bic_mcse=("mean_delta_bic_mcse", "median"),

        mean_nll_win_rate=("joint_nll_win_rate", "mean"),
        mean_bic_win_rate=("joint_bic_win_rate", "mean"),
    )
    .reset_index()
    .sort_values("particles")
)

particle_summary.round(4)
```

```python
df["joint_preferred_nll"] = df["mean_delta_nll"] > 0
df["joint_preferred_bic"] = df["mean_delta_bic"] > 0
```

```python
sign_summary = (
    df.groupby("particles")
    .agg(
        proportion_positive_delta_nll=("joint_preferred_nll", "mean"),
        proportion_positive_delta_bic=("joint_preferred_bic", "mean"),
    )
    .reset_index()
)

sign_summary
sign_summary[
    [
        "proportion_positive_delta_nll",
        "proportion_positive_delta_bic"
    ]
] *= 100

sign_summary
```

```python
setting_cols = [
    "mu_true",
    "alpha_true",
    "gamma_true",
    "phi_true",
    "r_true"
]

stability = (
    df.sort_values(setting_cols + ["particles"])
    [
        setting_cols
        + [
            "particles",
            "mean_delta_nll",
            "mean_delta_nll_mcse",
            "mean_delta_bic",
            "mean_delta_bic_mcse",
            "joint_nll_win_rate",
            "joint_bic_win_rate",
        ]
    ]
)

stability.round(4)
```
